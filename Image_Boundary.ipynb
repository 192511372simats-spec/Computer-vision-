```python
import cv2
import numpy as np
from google.colab import files
import matplotlib.pyplot as plt

f = files.upload()
img = cv2.imread(next(iter(f)), 0)

kernel = np.array([[-1,-1,-1],
                   [-1,8,-1],
                   [-1,-1,-1]])

boundary = cv2.filter2D(img, -1, kernel)

plt.imshow(boundary, cmap='gray')
plt.title("Image Boundary")
plt.axis("off")
plt.show()
```
