```python
# Import libraries
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()

# Get uploaded file name
image_name = list(uploaded.keys())[0]

# Read the image
image = cv2.imread(image_name)

# Check whether image is loaded
if image is None:
    print("Error: Image not found!")
else:
    # Get image dimensions
    height, width = image.shape[:2]

    # Bigger image (3x)
    scale_factor = 3.0
    bigger_image = cv2.resize(
        image,
        (int(width * scale_factor), int(height * scale_factor))
    )

    # Smaller image (0.5x)
    scale_factor = 0.5
    smaller_image = cv2.resize(
        image,
        (int(width * scale_factor), int(height * scale_factor))
    )

    # Display images
    print("Original Image")
    cv2_imshow(image)

    print("Bigger Image")
    cv2_imshow(bigger_image)

    print("Smaller Image")
    cv2_imshow(smaller_image)

    # Save images
    cv2.imwrite("Bigger_image.jpg", bigger_image)
    cv2.imwrite("Smaller_image.jpg", smaller_image)

    print("Images saved successfully!")
```

```python
# Import libraries
import cv2
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()

# Get uploaded file name
image_name = list(uploaded.keys())[0]

# Read the image
img = cv2.imread(image_name)

# Check whether image is loaded
if img is None:
    print("Error: Image not found!")
else:
    # Rotate clockwise (90 degrees)
    clockwise_img = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)

    # Rotate counterclockwise (90 degrees)
    counterclockwise_img = cv2.rotate(img, cv2.ROTATE_90_COUNTERCLOCKWISE)

    # Display images
    print("Original Image")
    cv2_imshow(img)

    print("Clockwise Rotated Image")
    cv2_imshow(clockwise_img)

    print("Counterclockwise Rotated Image")
    cv2_imshow(counterclockwise_img)

    # Save rotated images
    cv2.imwrite("Clockwise_Rotated_Image.jpg", clockwise_img)
    cv2.imwrite("Counterclockwise_Rotated_Image.jpg", counterclockwise_img)

    print("Images saved successfully!")
```

```python
# Import libraries
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()

# Get uploaded file name
image_name = list(uploaded.keys())[0]

# Read the image
image = cv2.imread(image_name)

# Check if image is loaded
if image is None:
    print("Error: Image not found!")
else:
    # Image dimensions
    rows, cols = image.shape[:2]

    # Translation values
    tx = 100   # Move 100 pixels to the right
    ty = 100   # Move 100 pixels downward

    # Translation matrix
    M = np.float32([[1, 0, tx],
                    [0, 1, ty]])

    # Translate the image
    moved_image = cv2.warpAffine(image, M, (cols, rows))

    # Display images
    print("Original Image")
    cv2_imshow(image)

    print("Moved Image")
    cv2_imshow(moved_image)

    # Save the translated image
    cv2.imwrite("Moved_Image.jpg", moved_image)

    print("Moved image saved successfully!")
```
