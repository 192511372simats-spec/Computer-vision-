```python
# 20. Sharpening of Image using Laplacian Mask with Negative Center Coefficient

import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

# Upload image
uploaded = files.upload()

# Get uploaded file name
image_path = next(iter(uploaded))

# Read image
img = cv2.imread(image_path)

# Convert to Grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Laplacian mask with negative center coefficient
laplacian_mask = np.array([[0, 1, 0],
                           [1, -4, 1],
                           [0, 1, 0]], dtype=np.float32)

# Apply Laplacian filter
laplacian = cv2.filter2D(gray, cv2.CV_32F, laplacian_mask)

# Sharpen the image
# For negative center coefficient:
# Sharpened Image = Original Image - Laplacian
sharpened = gray - laplacian

# Convert to 8-bit image
sharpened = np.clip(sharpened, 0, 255).astype(np.uint8)

# Display images
print("Original Image")
cv2_imshow(img)

print("Grayscale Image")
cv2_imshow(gray)

print("Laplacian Image")
cv2_imshow(cv2.convertScaleAbs(laplacian))

print("Sharpened Image")
cv2_imshow(sharpened)

# Save output
cv2.imwrite("Laplacian_Sharpened.jpg", sharpened)

# Download output
files.download("Laplacian_Sharpened.jpg")
```
