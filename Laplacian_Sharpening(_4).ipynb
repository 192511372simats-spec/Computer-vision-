```python
import cv2
import numpy as np
from google.colab import files
import matplotlib.pyplot as plt

uploaded = files.upload()
img = cv2.imread(next(iter(uploaded)))
img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

mask4 = np.array([[0,1,0],
                  [1,-4,1],
                  [0,1,0]])

lap4 = cv2.filter2D(img, -1, mask4)
sharp4 = cv2.add(img, -lap4)

plt.imshow(sharp4, cmap='gray')
plt.title("Laplacian Sharpening (-4)")
plt.axis("off")
plt.show()
```
