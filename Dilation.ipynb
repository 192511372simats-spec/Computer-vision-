```python
import cv2
from google.colab import files
import matplotlib.pyplot as plt

f = files.upload()
img = cv2.imread(next(iter(f)), 0)

kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (5,5))
result = cv2.dilate(img, kernel)

plt.imshow(result, cmap='gray')
plt.title("Dilation")
plt.axis("off")
plt.show()
```
