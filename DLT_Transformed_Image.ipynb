```python
# Direct Linear Transformation (DLT)

import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

# Get uploaded file name
image_path = next(iter(uploaded))

# Read image
image = cv2.imread(image_path)

# Convert BGR to RGB
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Get image size
rows, cols = image.shape[:2]

# Source points
src_pts = np.float32([
    [50, 50],
    [cols-50, 50],
    [50, rows-50],
    [cols-50, rows-50]
])

# Destination points
dst_pts = np.float32([
    [30, 100],
    [cols-100, 30],
    [100, rows-30],
    [cols-30, rows-100]
])

# Compute Homography Matrix using DLT
H, status = cv2.findHomography(src_pts, dst_pts, method=0)

# Apply Transformation
dlt_image = cv2.warpPerspective(image, H, (cols, rows))

# Display Images
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(dlt_image)
plt.title("DLT Transformed Image")
plt.axis("off")

plt.show()
```
