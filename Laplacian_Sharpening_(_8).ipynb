```python
import cv2
import numpy as np
from google.colab import files
import matplotlib.pyplot as plt

uploaded = files.upload()
img = cv2.imread(next(iter(uploaded)))
img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

mask8 = np.array([[1,1,1],
                  [1,-8,1],
                  [1,1,1]])

lap8 = cv2.filter2D(img, -1, mask8)
sharp8 = cv2.add(img, -lap8)

plt.imshow(sharp8, cmap='gray')
plt.title("Laplacian Sharpening (-8)")
plt.axis("off")
plt.show()
```
